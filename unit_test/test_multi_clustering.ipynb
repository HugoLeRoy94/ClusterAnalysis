```python
import numpy as np
import sys
import matplotlib.pyplot as plt
sys.path.append('../')
from src.embedding_base import EmbeddingBase
from src.cluster import Cluster
```

```python
# 1. Generate some data and create EmbeddingBase instances
Y1 = [np.array([[10, 10], [1.2, 1.3], [0.8, 0.9],[2,5],[0,8]])]
emb1 = EmbeddingBase(Y=Y1)
emb1.make_embedding(K=1)

Y2 = [np.array([[1, 1],[5.2, 4.8] , [9.8, 9.9]])]
emb2 = EmbeddingBase(Y=Y2)
emb2.make_embedding(K=1)

Y3 = [np.array([[5, 5], [10.2, 10.3], [4.8, 5.2]])]
emb3 = EmbeddingBase(Y=Y3)
emb3.make_embedding(K=1)
```

```python
# 2. Use the Cluster class to cluster the embeddings
n_clusters = 3
cluster_instance = Cluster(n_clusters=n_clusters)
cluster_instance.add_embeddings([emb1, emb2, emb3])
cluster_instance.set_shape_and_flatten()
cluster_instance.clusterize(n_clusters=n_clusters)
```

```python
# 3. Plot the results
plt.figure(figsize=(8, 6))

# Plot the cluster centers
plt.scatter(cluster_instance.cluster_centers_[:, 0], cluster_instance.cluster_centers_[:, 1],
            marker='*', s=200, c='black', label='Cluster Centers')

# Plot the points from each embedding with a different color
plt.scatter(emb1.flatten_embedding_matrix[:, 0], emb1.flatten_embedding_matrix[:, 1], c='r', label='Embedding 1')
plt.scatter(emb2.flatten_embedding_matrix[:, 0], emb2.flatten_embedding_matrix[:, 1], c='b', label='Embedding 2')
plt.scatter(emb3.flatten_embedding_matrix[:, 0], emb3.flatten_embedding_matrix[:, 1], c='g', label='Embedding 3')



plt.title('Clustering of Multiple Embeddings')
plt.xlabel('Dimension 1')
plt.ylabel('Dimension 2')
plt.legend()
plt.grid(True)
```
