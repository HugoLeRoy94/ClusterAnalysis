```python
# %load_ext autoreload
# %autoreload 2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import sys
sys.path.append('../../')

import src.viz

#sns.set(style="white", context="notebook")
```

```python
df_fed = pd.read_csv('../../data/choano/interim/entropy_scan_phases_fed_track.csv')
df_starve = pd.read_csv('../../data/choano/interim/entropy_scan_phases_starve_track.csv')
```

```python
fig,ax = plt.subplots(ncols=2,figsize=(14,7))

for K, group in df_fed.groupby("K"):
    ax[0].plot(group["n_clusters"], group["entropy_rate"], marker='o', label=f"K={K}")

ax[0].set_xlabel("Number of clusters ")
ax[0].set_ylabel("Entropy rate (bits)")
ax[0].set_title("Entropy rate vs number of clusters")
#ax[0].set_xlim(0,5)
ax[0].legend()
ax[0].set_xscale('log')

for K, group in df_starve.groupby("K"):
    ax[1].plot(group["n_clusters"], group["entropy_rate"], marker='o', label=f"K={K}")

ax[1].set_xlabel("Number of clusters ")
ax[1].set_ylabel("Entropy rate (bits)")
ax[1].set_title("Entropy rate vs number of clusters")
ax[1].set_ylim(0,0.2)
ax[1].legend()
ax[1].set_xscale('log')
```
